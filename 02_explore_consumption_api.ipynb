```python
import requests

url = "https://api.elhub.no/energy-data/v0/price-areas?dataset=CONSUMPTION_PER_GROUP_MBA_HOUR&startDate=2026-10-05&endDate=2026-10-06"
response = requests.get(url)
print(response.status_code)
```

```python
data = response.json()

for area in data["data"]:
    rows = area["attributes"]["consumptionPerGroupMbaHour"]
    print(area["id"], len(rows))
```

```python
rows = data["data"][1]["attributes"]["consumptionPerGroupMbaHour"]
print(rows[0])
```
