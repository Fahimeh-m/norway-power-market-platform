```python
import requests

url = "https://api.met.no/weatherapi/locationforecast/2.0/compact?lat=59.91&lon=10.75"
headers = {"User-Agent": "norway-power-market-platform github.com/Fahimeh-m"}

response = requests.get(url, headers=headers)
print(response.status_code)
```

```python
data = response.json()
steps = data["properties"]["timeseries"]

print(len(steps))
print(steps[0])
```

```python
places = {
    "NO1": (59.91, 10.75),  # Oslo
    "NO2": (58.15, 8.00),   # Kristiansand
    "NO3": (63.43, 10.40),  # Trondheim
    "NO4": (69.65, 18.96),  # Tromso
    "NO5": (60.39, 5.32),   # Bergen
}

for area, (lat, lon) in places.items():
    url = f"https://api.met.no/weatherapi/locationforecast/2.0/compact?lat={lat}&lon={lon}"
    response = requests.get(url, headers=headers)
    steps = response.json()["properties"]["timeseries"]
    first = steps[0]["data"]["instant"]["details"]
    print(area, len(steps), first["air_temperature"], first["wind_speed"])
```
